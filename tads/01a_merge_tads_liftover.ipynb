```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from concurrent.futures import ProcessPoolExecutor, as_completed
from tqdm import tqdm
```

```python
projdir = '/u/project/cluo/terencew/igvf/2023_YR2/snm3C/pseudobulk_hic'
clusters = pd.read_csv(f'{projdir}/txt/liftover.txt', header=None)[0].values
```

```python
tmp_cols = ['chrom', 'start', 'end', 'type', 'cluster']
```

```python
indir = f'{projdir}/tads/topdom/liftover'

chroms = [f'chr{x}' for x in range(1,23)]
final_list = []
for s in clusters:
    tads_list = []
    for chrom in chroms:
        tmp_tads = pd.read_csv(f'{indir}/{s}.{chrom}.bed', sep='\t', header=None)
        tads_list.append(tmp_tads)

    tads = pd.concat(tads_list)
    tads[1] = tads[1].astype(int)
    tads[2] = tads[2].astype(int)
    tads['cluster'] = s
    final_list.append(tads)

all_tads = pd.concat(final_list)
all_tads.columns = tmp_cols
all_tads.shape
```

```python
all_tads.to_csv(f'{projdir}/csv/tads/liftover.merged.csv', sep='\t')
```
